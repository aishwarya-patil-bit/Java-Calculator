A] Simple Linear Regressin by Using Sckit-learn

```python
#Feature selection
x=df[['Study_hours']] #Independent variables
y=df['Marks']         #Dependent variables
```

```python
x
```

Model Building

```python
#Split dataset into training and testing set
from sklearn.model_selection import train_test_split
x_train, x_test, y_train, y_test = train_test_split(x, y, test_size=0.3, random_state=42)
```

```python
x_test
```

```python
#Import Model
from sklearn.linear_model import LinearRegression
model = LinearRegression()

#Train the Model
model.fit(x_train, y_train)
```

```python
# predict the values
y_pred=model.predict(x_test)
y_pred
```

```python
# print equation
print('Intercept:', model.intercept_)
print('Slope:', model.coef_[0])
```

SLR Equation

 * Marks=(6.48*study_hours)+28.27

5) Model Evaluation

```python
#Evaluation
from sklearn.metrics import root_mean_squared_error, r2_score, mean_squared_error
rmse = np.sqrt(mean_squared_error(y_test, y_pred))
r2 = r2_score(y_test, y_pred)

print('RSME:', round(rmse, 3))
print('R2 score:', round(r2, 2))
```

```python
#Regression line

plt.figure(figsize=(7,5))
plt.scatter(x, y, color='blue', label='Actual data')
plt.plot(x_test, y_pred,
         color='red', label='Regression line')
plt.xlabel('Study Hours')
plt.ylabel('Marks')
plt.title('Simple Linear Regression')
plt.legend()
plt.show()
```

Residue Plot

```python
residuals = y_test - y_pred # Calculate Residual

plt.figure(figsize=(7,5))
plt.scatter(y_pred, residuals, color='green')
plt.axhline(y=0, color='red', linestyle='--')
plt.xlabel('Predicted Marks')
plt.ylabel('Residuals')
plt.title('Residue Plot')
plt.show()
```

B] Simple Linear Regression By Using Numpy


```python
# Independent Variable (features)
x = df["Study_hours"].values

#Dependent Variable (Target)
y = df["Marks"].values
```

```python
x
```

```python
# Calculate mean
x_mean = np.mean(x)
y_mean = np.mean(y)
```

```python
x_mean
```

```python
# Calculate Slope
m = np.sum((x-x_mean)*(y-y_mean))/np.sum((x-x_mean)**2)

#Calculate intercept
b = y_mean - m*x_mean

print("Slope =", m)
print("Intercept =", b)
```

```python
#Prediction
y_pred_numpy = m*x + b
y_pred_numpy
```
